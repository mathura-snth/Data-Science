```python
import numpy as np
from sklearn import datasets
import sklearn

iris = datasets.load_iris()

X = iris.data
Y = iris.target
```

## Plus proche voisin

```python
def PPV(X,Y):
    y_etiquette = []
    for i in range(len(X)):
        X_test = X[i].reshape(1,-1)

        X_train = np.delete(X,i,axis=0)
        Y_train = np.delete(Y,i)

        distances = sklearn.metrics.pairwise.euclidean_distances(X_train,X_test)

        voisin = np.argmin(distances)

        y_etiquette.append(Y_train[voisin])

    return np.array(y_etiquette)
PPV(X,Y)
```

### PPV avec renvoi de l'erreur de prédiction

```python
def PPV_err_pred(X,Y):
    y_etiquette = []
    compteur_erreur = 0
    for i in range(len(X)):
        X_test = X[i].reshape(1,-1)

        X_train = np.delete(X,i,axis=0)
        Y_train = np.delete(Y,i)

        distances = sklearn.metrics.pairwise.euclidean_distances(X_train,X_test)

        voisin = np.argmin(distances)

        y_etiquette.append(Y_train[voisin])

    for k in range(len(y_etiquette)):
        if(y_etiquette[k] != Y[k]):
            compteur_erreur += 1

    pourcentage_erreur = (compteur_erreur/len(Y)) * 100

    return np.array(y_etiquette) , pourcentage_erreur
```

### Tests sur les données Iris

```python
etiquette, pourcentage = PPV_err_pred(X,Y)
print(etiquette)
print(f"Pourcentage d'étiquettes mal prédites : {pourcentage}")
```

### Test de KNeighborsClassifier avec différentes K

```python
def KNN(k):

    K_neigh = sklearn.neighbors.KNeighborsClassifier(n_neighbors = k)
    K_neigh.fit(X, Y)

    y_pred_knn = K_neigh.predict(X)

    erreur_knn = (np.sum(y_pred_knn != Y) / len(Y)) * 100

    print("Prédictions Sklearn :", y_pred_knn)
    print(f"Erreur : {erreur_knn}%")

KNN(1)
KNN(3)
```

On remarque que les résultats sont différents selon le nombre de voisins que l'on prend en considération, d'ailleurs on remarque aussi que pour K = 1 avec Sklearn, puisque les données d'entrainements sont les mêmes que les données de tests, on est inclu dans nos voisins donc 1-nn revient à soi, d'où l'erreur de prédiction de 0%, contrairement à la fonction PPV où l'on s'exclue (delete).

## Classifieur Bayesien Naïf

```python
def CBN(X,Y):
    classe = np.unique(Y)
    proba_priori = []
    barycentre = []
    for c in classe:
        X_type_c = []
        for i in range(len(X)):
            if Y[i] == c :
                X_type_c.append(X[i])
        proba_priori.append(len(X_type_c) / len(X))
        barycentre.append(np.mean(X_type_c,axis=0))
    print(proba_priori)
    print(barycentre)

    CBN_res = []
    for x in X:
        res = []
        for k in range(len(classe)):
            proba_x_sachant_w = []

            for i in range(len(x)):
                xi = x[i]

                distances = []
                for bary in barycentre : 
                    distances.append(abs(xi - bary[i]))

                D = np.sum(distances)

                if D!=0 :
                    proba_x_sachant_w.append(1-(distances[k] / D))
                else :
                    proba_x_sachant_w.append(1.0 / len(classe))

            produit = np.prod(proba_x_sachant_w) * proba_priori[k]
            res.append(produit)

        maximum = np.argmax(res)
        CBN_res.append(maximum)

    return np.asarray(CBN_res)

l = CBN(X,Y)
print(l)
```

### Test naive_bayes.GaussianNB

```python
def gaussiannb() :
    gaussiannb = sklearn.naive_bayes.GaussianNB()
    gaussiannb.fit(X,Y)

    y_pred_gaussiannb = gaussiannb.predict(X)

    err_gaussiannb = (np.sum(y_pred_gaussiannb != Y) / len(Y)) * 100

    print("Prédictions Sklearn :", y_pred_gaussiannb)
    print(f"Erreur : {err_gaussiannb}%")

gaussiannb()
```

Nous remarquons que les résultats sont différents, 12,67% pour la fonction définie manuellement contre 4% pour celle de Sklearn. Cela est dû au fait que nous avons uniquement calculé la distance aux barycentres dans CBN alors que GaussianNB utilise la moyenne mais aussi l'écart-type.

```python
def CBN_proba(X,Y):
    classe = np.unique(Y)
    proba_priori = []
    barycentre = []
    variance = []
    for c in classe:
        X_type_c = []
        for i in range(len(X)):
            if Y[i] == c :
                X_type_c.append(X[i])
        proba_priori.append(len(X_type_c) / len(X))
        barycentre.append(np.mean(X_type_c,axis=0))
        variance.append(np.var(X_type_c,axis=0))

    print(proba_priori)
    print(barycentre)
    print(variance)

    CBN_res = []
    for x in X:
        res = []
        for k in range(len(classe)):
            var = variance[k]
            mu = barycentre[k]

            proba_x_sachant_w = (
                1 / np.sqrt(2 * np.pi * var)
                * np.exp(-((x - mu) ** 2) / (2 * var))
            )

            produit = np.prod(proba_x_sachant_w) * proba_priori[k]
            res.append(produit)

        maximum = np.argmax(res)
        CBN_res.append(maximum)

    return np.asarray(CBN_res)

l = CBN_proba(X,Y)
print(l)
```
